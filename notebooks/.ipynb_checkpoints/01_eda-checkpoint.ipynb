```python
import pandas as pd

# Загружаем данные. Убедитесь, что путь верный.
df = pd.read_csv('../data/raw/cs-training.csv')

# Показываем первые 5 строк, чтобы увидеть структуру
print("Первые 5 строк данных:")
print(df.head())

# Получаем общую информацию о данных
print("\nИнформация о данных:")
print(df.info())

# Получаем статистическое описание числовых колонок
print("\nСтатистическое описание:")
print(df.describe())
```

```python
# Посчитаем количество значений для каждого класса
target_counts = df['SeriousDlqin2yrs'].value_counts()
print("Количество значений в целевой переменной:")
print(target_counts)

# Визуализация распределения
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.countplot(x='SeriousDlqin2yrs', data=df)
plt.title('Распределение целевой переменной')
plt.show()

# Рассчитаем процентное соотношение
print("\nПроцентное соотношение классов:")
print(target_counts / len(df) * 100)
```

```python
# Анализ возраста
print("\nСтатистика по возрасту:")
print(df['age'].describe())

# Визуализация распределения возраста
plt.figure(figsize=(10, 6))
sns.histplot(df['age'], kde=True)
plt.title('Распределение возраста')
plt.show()
```

```python
# Поиск пропусков во всех колонках
missing_values = df.isnull().sum()
print("\nКоличество пропущенных значений в каждой колонке:")
print(missing_values[missing_values > 0])
```

```python
# Связь возраста с просрочкой
plt.figure(figsize=(10, 6))
sns.boxplot(x='SeriousDlqin2yrs', y='age', data=df)
plt.title('Зависимость просрочки от возраста')
plt.show()

# Связь MonthlyIncome с просрочкой (для тех, у кого нет пропусков)
plt.figure(figsize=(10, 6))
sns.boxplot(x='SeriousDlqin2yrs', y='MonthlyIncome', data=df.dropna(subset=['MonthlyIncome']))
plt.title('Зависимость просрочки от ежемесячного дохода')
plt.ylim(0, 20000) # Ограничим ось, чтобы убрать выбросы
plt.show()
```
